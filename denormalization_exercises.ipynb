# Denormalization Exercises

## 1

A normalized data schema in a retail firm contains the following tables:

```
Users(user_id, name, email)

Orders(order_id, user_id, order_date, total)

OrderItems(order_item_id, order_id, product_id, quantity, price)

Products(product_id, name, category)
```

a) Explain the main disadvantage of this schema.

b) Design a denormalized schema optimized for a dashboard that can display all information related to an order:
- user name
- order date
- total price
- products contained in the order (quantity and price)

c) Give some example rows in your table.

## 2

We now have the following normalized schema for a dashboard of healthcare appointments:

```
Patients(patient_id, name, date_of_birth)

Doctors(doctor_id, name, specialty)

Appointments(appointment_id, patient_id, doctor_id, appointment_time)

Diagnoses(diagnosis_id, appointment_id, code, description)

Prescriptions(prescription_id, appointment_id, medication_name, dosage)
```

Assume each appointment results in, at most, one diagnosis and at most one prescription.

a) Design a single-table denormalized schema optimized for a dashboard that shows all interactions between a patient and doctor:
- patient name
- doctor name
- appointment time
- list of diagnoses (codes + descriptions)
- list of prescribed medications and their dosage

b) Write a few example rows of your table.

c) What could serve as a primary key?
